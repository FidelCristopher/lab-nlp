```python
from nltk.tokenize import sent_tokenize, word_tokenize
from gensim.models import Word2Vec
```

```python
with open("alice.txt", "r", encoding="utf-8") as file:
    temp = file.read()
doc = temp.replace("\n", " ")
```

```python
data = []

# [[], [], []]

for sentence in sent_tokenize(doc):
    tokens = []
    for token in word_tokenize(sentence):
        tokens.append(token.lower())
        
    data.append(tokens)
```

```python
skipgram = Word2Vec(data, min_count=1, vector_size=100 , sg=1, window=5, workers=4)
cbow = Word2Vec(data, min_count=1, window=5, workers=4)
```

```python
skipgram.wv.most_similar("alice", topn=100)
```

```python
skipgram.wv.similarity('alice', 'wonderland')
```

```python
cbow.wv.most_similar("alice", topn=100)
```

```python
cbow.wv.similarity('alice', 'wonderland')
```

## GLOVE

```python
import numpy as np
```

```python
def load_glove(path):
    glove_dict = {}
    
    with open(path, "r", encoding="utf-8") as file:
        for line in file:
            values = line.split(" ")
            word = values[0]
            vector = np.asarray(values[1:], dtype="float32")
            glove_dict[word] = vector
            
    return glove_dict
```

```python
glove_path = "glove.6B.100d.txt"
glove_dict = load_glove(glove_path)
```

```python
glove_dict["king"]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
from scipy.spatial.distance import cosine
```

```python
word_1 = "king"
word_2 = "queen"

if word_1 in glove_dict.keys() and word_2 in glove_dict.keys():
    sin = cosine_similarity([glove_dict[word_1]], [glove_dict[word_2]])
    print(sin)
```
