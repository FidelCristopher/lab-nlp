```python
from nltk.stem import LancasterStemmer, PorterStemmer, SnowballStemmer, WordNetLemmatizer
```

```python
word_list = ['Writing', 'playing', 'played', 'riding', 'drives', 'breaking', 'lives', 'running', 'given', 'chosen', 'drawing']
```

```python
lancaster = LancasterStemmer()
porter = PorterStemmer()
snowball = SnowballStemmer('english')
```

```python
for word in word_list:
    print(f'aslinya = {word}')
    print("==================")
    print(f"Lancaster = {lancaster.stem(word)}")
    print(f"Porter = {porter.stem(word)}")
    print(f"Snowball = {snowball.stem(word)}")
    print(" ")
```

```python
lemmatizing = WordNetLemmatizer()
```

```python
kata = 'better'

verb = lemmatizing.lemmatize(kata, 'v')
adj = lemmatizing.lemmatize(kata, 'a')
adv = lemmatizing.lemmatize(kata, 'r')
noun = lemmatizing.lemmatize(kata, 'n')

print(verb, adj, adv, noun)
```

```python
from nltk.tokenize import word_tokenize
```

```python
sentence = 'Quickly kicking quirky kangaroos caused chaos'
list_token = word_tokenize(sentence)
list_token
```

```python
for token in list_token:
    synonyms = []
    antonyms = []
    synsets = wordnet.synsets(token)
    
    for syn in synsets:
        lemmas = syn.lemmas()

        for lemma in lemmas:
            synonym = lemma.name()
            synonyms.append(synonym)

            for ant in lemma.antonyms():
                antonym = ant.name()
                antonyms.append(antonym)
                
    print(token)
    print("=============")
    print(f"synonyms = {synonyms}")
    print(f"antonyms = {antonyms}")
    print(" ")
```
