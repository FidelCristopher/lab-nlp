## POS TAGGING

```python
from nltk.tag import pos_tag
from nltk.tokenize import word_tokenize
import string 
from nltk.corpus import stopwords
```

```python
sentence = "Four frantic fishermen named Bobi fled from the freezing fjord after finding forty-five flimsy fish fluttering frantically in a fragile, floating framework that finished falling forward into the foam."

eng_stopwords = stopwords.words('english')
punctuation = string.punctuation

word_list = word_tokenize(sentence)
# print(word_list)
remove_stopwords = [word for word in word_list if word.lower() not in eng_stopwords]
# print(remove_stopwords)
remove_punc = [token for token in remove_stopwords if token not in punctuation]
# print(remove_punc)
```

```python
tagged = pos_tag(remove_punc)
print(tagged)
```

```python
from nltk.chunk import ne_chunk
```

```python
ne = ne_chunk(tagged)
print(ne)
```

## FREQUENCY DISTRIBUTION 

```python
from nltk.probability import FreqDist
```

```python
fd = FreqDist(word_list)
for word, count in fd.most_common():
    print(f'{word} : {count}')
```

## LATIHAN

```python
sentence = "A dedicated doctor declared that a dozen dormant dogs deserved distinct doses of double-strength drops to decrease their drastic discomfort, but the dogs decided to dodge the doctor’s daunting demands instead."

eng_stopwords = stopwords.words('english')
punctuation = string.punctuation

word_list = word_tokenize(sentence)
print(word_list)

remove_stopwords = [word for word in word_list if word.lower() not in eng_stopwords]
print(remove_stopwords)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = remove_stopwords
stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

words = remove_stopwords
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
print(lemmatized_words)
```

```python
tagged = pos_tag(remove_punc)
print(tagged)

ne = ne_chunk(tagged)
print(ne)
```

```python
fd = FreqDist(tagged)
for word, count in fd.most_common():
    print(f'{word} : {count}')
```
