```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string
import nltk
nltk.download('punk_tab')
nltk.download('stopwords')
```

```python
while 1:

    sentence = input("Input sentence(s): ")
    
    if sentence.strip() == "":
        continue
    
    tokenize = word_tokenize(sentence)
    print(tokenize)

    break
```

```python
stop_words = stopwords.words('english')

remove_stop = [word for word in tokenize if word in stop_words]
print(remove_stop)
```

```python
from nltk.stem import SnowballStemmer
stemmer = SnowballStemmer('english')

print("Your stemmed words are:")
print(" ")
for word in tokenize:
    print(f"{word} -> {stemmer.stem(word)}")
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

print("Your lemmatized words are: ")
print(" ")
for word in tokenize:
    print(f" {word} -> {lemmatizer.lemmatize(word)}")
```

```python
from nltk.tag import pos_tag
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
tagged = pos_tag(tokenize)
print(tagged)
```

```python
from nltk.chunk import ne_chunk

ne = ne_chunk(tagged)
ne
```

```python
from nltk.probability import FreqDist

fd = FreqDist(tokenize)
for word, count in fd.most_common():
    print(f"{word} = {count}")
```
