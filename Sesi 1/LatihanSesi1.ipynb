```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
paragraph = "Peter plays The pleasant piano pieces. Professional players praise Peter because Peter performs perfectly. Big brown bears bake bright berry biscuits. Brave birds bring black beans back. Busy bees build better boxes behind Blue Bay."

sentence = sent_tokenize(paragraph)
print(sentence)

word_list = word_tokenize(paragraph)
print(word_list)
```

```python
from nltk.corpus import stopwords
stop_words = stopwords.words('english')

removed_stopwords = [word for word in word_list if word.lower() not in stop_words]
print(f"before remove: {word_list}")
print(f"after remove: {removed_stopwords}")
```

```python
import string
punctuations = string.punctuation

removed_punctuations = [word for word in removed_stopwords if word not in punctuations]
print(f"before remove punctuation: {removed_stopwords}")
print(f"after removed punctuations: {removed_punctuations}")
```
