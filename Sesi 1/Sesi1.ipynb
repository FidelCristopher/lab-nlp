## TOKENIZING

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
sentence = "Many merry monkeys make messy meals. The marry monkeys makes Mr. Moon may manage many more moves."

word_list = word_tokenize(sentence)
print(word_list)
```

```python
paragraph = "Peter plays pleasant piano pieces. Professional players praise Peter because Peter performs perfectly. Big brown bears bake bright berry biscuits. Brave birds bring black beans back. Busy bees build better boxes behind Blue Bay."

sentence_list = sent_tokenize(paragraph)
print(sentence_list)
```
